```python
## tkinter initialization

from tkinter import *
import tkinter.ttk as ttk
import tkinter. messagebox as tkMessageBox
import sqlite3

root = Tk()
root.geometry('1000x400')
root.title('Library Book Management System')
```

```python
## methods

def Database():
    global conn,cursor
    conn = sqlite3.connect('library.db')
    cursor = conn.cursor()
    cursor.execute('CREATE TABLE IF NOT EXISTS booklist(booktitle TEXT,author TEXT,year INTEGER,isbn TEXT)')
    
def Add():
    if book_title.get() == '' or author.get() =='' or year.get() == '':
        txt_result.config(text = 'Please enter all the fields. ISBN is optional', fg='red')
    else:
        Database()
        cursor.execute('INSERT INTO booklist(booktitle,author,year,isbn) VALUES(?,?,?,?)',(book_title.get(),author.get()
                                                        ,year.get(),isbn.get()))
        conn.commit()
        book_title.set('')
        author.set('')
        year.set('')
        isbn.set('')
        cursor.close()
        conn.close()
        txt_result.config(text = 'Book Record Inserted',fg='green')
        
def Issue():
    txt_result.config(text = 'The Book has been issued', fg = 'black')

def Delete():
    Database()
    bt = book_title.get()
    isbn_dummy = isbn.get()
    cursor.execute('DELETE FROM booklist WHERE booktitle = ? or isbn = ? ',(bt,isbn_dummy))
    conn.commit()
    cursor.close()
    conn.close()
    txt_result.config(text = 'Book Record Deleted', fg = 'cyan')
    book_title.set('')
    author.set('')
    year.set('')
    isbn.set('')

def Display():
    tree.delete(*tree.get_children())
    Database()
    cursor.execute('SELECT * FROM booklist ORDER BY booktitle ASC')
    fetch = cursor.fetchall()
    for data in fetch:
        tree.insert('','end',values =(data[0],data[1],data[2],data[3]) )
    cursor.close()
    txt_result.config(text = 'Sucessfully displayed data on the tree',fg = 'brown')
    book_title.set('')
    author.set('')
    year.set('')
    isbn.set('')
    
    
def Search():
    tree.delete(*tree.get_children())
    Database()
    bt = book_title.get()
    isbn_dummy = isbn.get()
    cursor.execute('SELECT * FROM booklist WHERE booktitle = ? or isbn = ?',(bt,isbn_dummy))
    fetch = cursor.fetchall()
    for data in fetch:
        tree.insert('','end',values =(data[0],data[1],data[2],data[3]) )
    cursor.close()
    txt_result.config(text = 'Search Complete',fg = 'orange')
    book_title.set('')
    author.set('')
    year.set('')
    isbn.set('')
    
def Exit():
    result = tkMessageBox.askquestion('Do you want to exit ? (y/n)',icon='warning')
    if result == 'yes':
        root.destroy()
        exit()

def del_Database():
    Database()
    cursor.execute('DROP TABLE booklist')
```

```python
## variable declaration

book_title = StringVar()
author = StringVar()
year = IntVar()
year.set('')
isbn = StringVar()
```

```python
## Frame

Top = Frame(root,width = 900,height = 50, bd = 8,background="blue")
Top.pack(side = TOP)
Left = Frame(root,width = 300,height = 500, bd = 8, relief = 'raise', bg = 'red')

Left.pack(side = LEFT)
Right = Frame(root,width = 600,height = 500, bd = 8, relief = 'raise', bg = 'yellow')
Right.pack(side = RIGHT)

Forms = Frame(Left,width = 300,height = 400)
Forms.pack(side = TOP)

Buttons = Frame(Left, width = 300,height = 100)
Buttons.pack(side = BOTTOM)
```

```python
## Labels

txt_title = Label(Top,width = 900,font=('arial',28),text = 'Library Book Management System')
txt_title.pack()
txt_book_title = Label(Forms, text = 'Book Title', font = ('arial',16),bd=15 )
txt_book_title.grid(row = 0, stick = 'e')
txt_author = Label(Forms, text = 'Author', font = ('arial',16),bd=15 )
txt_author.grid(row = 1, stick = 'e')
txt_year = Label(Forms, text = 'Year', font = ('arial',16),bd=15 )
txt_year.grid(row = 2, stick = 'e')
txt_isbn = Label(Forms, text = 'ISBN', font = ('arial',16),bd=15 )
txt_isbn.grid(row = 3, stick = 'e')

txt_result = Label(Buttons, text = '',width = 40)
txt_result.pack(side = TOP)
```

```python
## Entries

booktitle_entry = Entry(Forms, textvariable = book_title, width = 30)
booktitle_entry.grid(row = 0, column = 1)
author_entry = Entry(Forms, textvariable = author,width = 30)
author_entry.grid(row = 1, column =1)
year_entry = Entry(Forms, textvariable = year, width = 30)
year_entry.grid(row = 2, column = 1)
isbn_entry = Entry(Forms, textvariable = isbn, width = 30)
isbn_entry.grid(row = 3, column = 1)
```

```python
## buttons
btn_display = Button(Buttons,width = 10, text = 'Display', command = Display)
btn_display.pack(side = LEFT)
btn_search = Button(Buttons,width = 10, text = 'Search', command = Search)
btn_search.pack(side = LEFT)
btn_add = Button(Buttons,width = 10, text = 'Add', command = Add)
btn_add.pack(side = LEFT)
btn_issue = Button(Buttons,width = 10, text = 'Issue', command = Issue)
btn_issue.pack(side = LEFT)
btn_delete = Button(Buttons,width = 10, text = 'Delete', command = Delete)
btn_delete.pack(side = LEFT)
btn_exit = Button(Buttons,width = 10, text = 'Exit', command = Exit)
btn_exit.pack(side = LEFT)
```

```python
## list Widget

scrollbarx = Scrollbar(Right, orient = VERTICAL)
scrollbary = Scrollbar(Right, orient = HORIZONTAL)

tree = ttk.Treeview(Right, columns=('Book Title','Author','Year','ISBN'))

scrollbary.config(command=tree.yview)
scrollbary.pack(side = RIGHT, fill = Y)
scrollbarx.config(command=tree.xview)
scrollbarx.pack(side = BOTTOM, fill = X)

tree.heading('Book Title', text = 'BookTitle', anchor = W)
tree.heading('Author', text = 'Author', anchor = W)
tree.heading('Year', text = 'Year', anchor = W)
tree.heading('ISBN', text = 'ISBN', anchor = W)

tree.column('#0',stretch = NO,minwidth = 0,width=0)
tree.column('#1',stretch = NO,minwidth = 0,width=100)
tree.column('#2',stretch = NO,minwidth = 0,width=150)
tree.column('#3',stretch = NO,minwidth = 0,width=50)

tree.pack()




```

```python
## main loop
if __name__ =='__main__':
    root.mainloop()
```
