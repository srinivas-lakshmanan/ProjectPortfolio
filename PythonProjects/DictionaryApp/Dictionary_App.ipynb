```python

import json

## Main menu
isrun = 1

while(isrun):
    print('''
    Main Menu

    Add a new word ---- 1
    Find the meaning ---- 2
    Update a word --- 3
    Exit ---- 4

    ''')
    


    
    choice = int(input('Select an option from the main menu using the numbers: \n '))
    
    if choice == 1:
        temp_word = input('Enter the new word \n').lower()
        temp_meaning = input('Enter the meaning \n')
        temp_dict = {temp_word:temp_meaning}
        with open('words.txt','r') as read_file:
            data= read_file.read()
            js_data = json.loads(data)
        with open('words.txt','w') as addition_file:
            js_data.update(temp_dict)
            addition_file.write(json.dumps(js_data))
    
    if choice == 2:
        temp_word = input('Enter the word whose meaning you would like to look-up: \n ').lower()
        with open('words.txt','r') as read_file:
            data= read_file.read()
            js_data = json.loads(data)
        print('The meaning of the word ',temp_word,' is: ',js_data[temp_word])
    
    if choice == 3:
        temp_word = input('Enter the word whose meaning you would like to update: \n').lower()
        temp_meaning = input('Enter the new meaning of the word: \n')
        with open('words.txt','r') as read_file:
            data= read_file.read()
            js_data = json.loads(data)
        
        with open('words.txt','w') as update_file:
            js_data[temp_word]=temp_meaning
            update_file.write(json.dumps(js_data))
    
    if choice == 4:
        print('Closing the dictionary app...')
        isrun = 0
        
    if not choice == 1 and not choice == 2 and not choice ==3 and not choice == 4 :
        print('Enter a valid choice(1 or 2 or 3 or 4)....\n')
        


```
